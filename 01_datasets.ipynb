## 1. Samson dataset

The **Samson** dataset is a widely used benchmark for hyperspectral unmixing. It was acquired by the HYDICE (Hyperspectral Digital Imagery Collection Experiment) sensor over the Samson area. 

The hyperspectral image consists of a scene with 95 × 95 pixels and 156 spectral bands, extracted from a larger image and covering wavelengths ranging from 401 nm to 889 nm (spectral resolution: 3.13 nm). 

The dataset contains three dominant endmembers (``soil``, ``tree``, and ``water``) and is commonly used to evaluate spectral unmixing algorithms due to its relatively simple composition and available reference abundance maps. Its moderate size and well-characterized materials make it a standard benchmark for comparing both linear and nonlinear unmixing methods.

```python
HSI_path = "src/dataset/samson.npz"
HSI = HyperspectralData.load_npz(HSI_path)
```

### Endmembers and abundances map

The following plots the endmembers along with the corresponding abundance maps.

```python
plotter = HyperspectralPlotter(HSI)
fig1 = plotter.plot_endmembers(figsize=(9, 3))
fig2 = plotter.plot_abundances(figsize=(10, 3))
```

Finally, it is possible to export the dataset as ``.mat`` files. The keys are :

- ``wavelengths``,
- ``wavelength_unit``
- ``image_shape``,
- ``endmember_names``,

- ``per_pixel_endmembers``.

## 2. Jasper Ridge dataset

The Jasper Ridge dataset is another widely used benchmark for hyperspectral unmixing. It was acquired by the Airborne Visible/Infrared Imaging Spectrometer (AVIRIS) over the Jasper Ridge Biological Preserve in California, USA.

The original hyperspectral image contains 224 spectral bands covering wavelengths from 380 nm to 2500 nm, with a spectral resolution of approximately 9.46 nm. The original image has 512 × 614 pixels with a ground sample distance of 20 m. 

The benchmark dataset commonly used in the literature consists of a 100 × 100 pixel subscene with 198 spectral bands, obtained after removing noisy and water absorption bands (channels 1–3, 108–112, 154–166, and 220–224).

The scene contains four reference endmembers: `Road`, `Soil`, `Water`, `Tree`. Ground-truth abundance maps for these four materials are also provided, making the dataset a standard benchmark for evaluating hyperspectral unmixing algorithms.

```python
HSI_path = "src/dataset/jasper_ridge.npz"
HSI = HyperspectralData.load_npz(HSI_path)
plotter = HyperspectralPlotter(HSI)
fig1 = plotter.plot_endmembers()
fig2 = plotter.plot_abundances()
```

## 3. Urban dataset

Urban is a 307x307 pixels image, each of which corresponds to a 2m $\times $ 2m area. The data set was acquired by imaging an area of Michigan, USA, with the Hyperspectral Digital Imagery Collection Experiment (HYDICE) sensor. 

210 wavelengths ranging from 400 nm to 2500 nm are considered, resulting in a spectral resolution of 10 nm. The channels 1-4, 76, 87, 101-111, 136-153 and 198-210 are removed due to dense water vapor and atmospheric effects, resulting in 162 spectral bands overall. 

There are three versions of ground truth, which contain 4, 5 and 6 endmembers respectively.

### Urban 4 dataset

For Urban 4, the endmembers are `asphalt`, `grass`, `tree` and `roof`.

```python
HSI_path = "src/dataset/urban4.npz"
HSI = HyperspectralData.load_npz(HSI_path)
plotter = HyperspectralPlotter(HSI)
fig1 = plotter.plot_endmembers(figsize=(10, 3))
fig2 = plotter.plot_abundances(figsize=(11, 3))
```

## 4. Apex dataset

Apex is a 110x110 pixels image, with 4 endmembers: `water`, `tree`, `road`, and `roof`. 285 wavelengths ranging from 413 nm to 2420 nm are considered.

## 5. Cuprite dataset

Cuprite is a challenging hyperspectral image captured by the AVIRIS sensor. It covers an area near Las Vegas, Nevada, USA. 

The image contains 224 channels, ranging from 370 nm to 2480 nm. The noisy (1-2 and 221-224) and water absorption channels (104-113 and 148-167) are usually removed, yielding a total of 188 channels. A subimage of 250 x 190 pixels is usually considered, which includes 14 types of minerals. 

Since there are only minor spectral differences between variants of some minerals, the number of endmembers is reduced to 12 in practice, summarized as follows : ``#1 Alunite``, ``#2 Andradite``, ``#3 Buddingtonite``, ``#4 Dumortierite``, ``#5 Kaolinite1``, ``#6 Kaolinite2``, ``#7 Muscovite``, ``#8 Montmorillonite``, ``#9 Nontronite``, ``#10 Pyrope``, ``#11 Sphene``, ``#12 Chalcedony``.

Note that no ground-truth abundance maps are available for Cuprite.

```python
HSI_path = "src/dataset/cuprite.npz"
HSI = HyperspectralData.load_npz(HSI_path)
Y = HSI.observations
plotter = HyperspectralPlotter(HSI)
fig1 = plotter.plot_endmembers(
    figsize=(10, 6),
    max_cols=4,
)
```

## 6. Washington DC Mall

The Washington DC Mall (WDM) is an airborne hyperspectral image flightline over the Washington DC Mall, collected by the HYDICE sensor.

The subimage considered here has 319 × 292 pixels and 191 spectral bands, covering wavelengths from 400 nm to about 2360 nm after removal of the water absorption bands.

The scene contains six reference endmembers: `Roof`, `Grass`, `Water`, `Tree`, `Trail` and `Road`. Ground-truth abundance maps are also provided.

```python
HSI_path = "src/dataset/WDM.npz"
HSI = HyperspectralData.load_npz(HSI_path)
Y = HSI.observations
plotter = HyperspectralPlotter(HSI)
fig1 = plotter.plot_endmembers(
    figsize=(10, 5),
    max_cols=3,
)

fig2 = plotter.plot_abundances(figsize=(13, 3))
```
